## 7. Data Cleaning and Preprocessing

Data cleaning is an important step in preparing the agricultural dataset for reliable analysis.
The dataset was examined for missing values, duplicate records, incorrect data types, and
potential inconsistencies before performing further analysis.

```python
# Creating a copy of the original dataset for cleaning

df_clean = df.copy()

print("Original dataset shape:", df_clean.shape)
```

```python
# Handle missing values

# Rainfall and soil moisture: fill with overall median
df_clean["Rainfall_mm"] = df_clean["Rainfall_mm"].fillna(
    df_clean["Rainfall_mm"].median()
)

df_clean["Soil_Moisture_pct"] = df_clean["Soil_Moisture_pct"].fillna(
    df_clean["Soil_Moisture_pct"].median()
)

# Yield: fill using Crop + Season group median
df_clean["Yield_Tonnes_Ha"] = df_clean.groupby(
    ["Crop", "Season"]
)["Yield_Tonnes_Ha"].transform(
    lambda x: x.fillna(x.median())
)

print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# Checking and removing completely duplicated records

duplicates_before = df_clean.duplicated().sum()

print("Duplicate records found:", duplicates_before)

if duplicates_before > 0:
    df_clean = df_clean.drop_duplicates()

print("Dataset shape after duplicate handling:", df_clean.shape)
```

```python
# Final data quality check

print("Final dataset shape:", df_clean.shape)
print("Total missing values:", df_clean.isnull().sum().sum())
print("Total duplicate rows:", df_clean.duplicated().sum())
```

## 14. Statistical Analysis of Seasonal Yield Differences

Statistical testing is performed to determine whether the observed differences in
agricultural yield between seasons are statistically significant.

A one-way ANOVA is used to compare the mean yield across Kharif, Rabi, and Zaid seasons.
A significance level of 5% is used for the hypothesis test.

```python
# One-way ANOVA: Yield across seasons

from scipy.stats import f_oneway

kharif_yield = df_clean[df_clean["Season"] == "Kharif"]["Yield_Tonnes_Ha"]
rabi_yield = df_clean[df_clean["Season"] == "Rabi"]["Yield_Tonnes_Ha"]
zaid_yield = df_clean[df_clean["Season"] == "Zaid"]["Yield_Tonnes_Ha"]

anova_stat, anova_p = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("One-Way ANOVA Results")
print("---------------------")
print(f"F-statistic: {anova_stat:.4f}")
print(f"p-value: {anova_p:.6f}")

if anova_p < 0.05:
    print("\nResult: The difference in average yield across seasons is statistically significant.")
else:
    print("\nResult: The difference in average yield across seasons is not statistically significant.")
```

```python
# Calculate eta-squared effect size for seasonal yield

grand_mean = df_clean["Yield_Tonnes_Ha"].mean()

between_group_ss = sum(
    len(group) * (group["Yield_Tonnes_Ha"].mean() - grand_mean) ** 2
    for _, group in df_clean.groupby("Season")
)

total_ss = sum(
    (df_clean["Yield_Tonnes_Ha"] - grand_mean) ** 2
)

eta_squared = between_group_ss / total_ss

print(f"Eta-squared (η²): {eta_squared:.4f}")
```

```python
# Kruskal-Wallis test: Yield across seasons

from scipy.stats import kruskal

kruskal_stat, kruskal_p = kruskal(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("Kruskal-Wallis Test Results")
print("---------------------------")
print(f"H-statistic: {kruskal_stat:.4f}")
print(f"p-value: {kruskal_p:.6f}")

if kruskal_p < 0.05:
    print("\nResult: Yield distributions differ significantly across seasons.")
else:
    print("\nResult: No statistically significant difference was detected across seasons.")
```

```python
# Distribution of yield across seasons

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df_clean,
    x="Season",
    y="Yield_Tonnes_Ha"
)

plt.title("Yield Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
# Detailed yield statistics by season

yield_statistics = df_clean.groupby("Season")["Yield_Tonnes_Ha"].agg(
    Count="count",
    Mean="mean",
    Median="median",
    Standard_Deviation="std",
    Minimum="min",
    Maximum="max"
).round(3)

yield_statistics
```

```python
# Pairwise Mann-Whitney U tests

from scipy.stats import mannwhitneyu
from itertools import combinations

seasons = df_clean["Season"].unique()

results = []

for season1, season2 in combinations(seasons, 2):

    group1 = df_clean[df_clean["Season"] == season1]["Yield_Tonnes_Ha"]
    group2 = df_clean[df_clean["Season"] == season2]["Yield_Tonnes_Ha"]

    statistic, p_value = mannwhitneyu(
        group1,
        group2,
        alternative="two-sided"
    )

    # Bonferroni correction for 3 pairwise comparisons
    adjusted_p = min(p_value * 3, 1.0)

    results.append({
        "Comparison": f"{season1} vs {season2}",
        "U_Statistic": statistic,
        "Raw_p_value": p_value,
        "Adjusted_p_value": adjusted_p
    })

pairwise_results = pd.DataFrame(results)

pairwise_results.round(6)
```

### Pairwise Seasonal Yield Comparison

Pairwise Mann–Whitney U tests were conducted to identify which specific seasons differed in yield distribution. Bonferroni correction was applied to account for multiple comparisons.

All three pairwise comparisons produced adjusted p-values below the 5% significance level:

- Kharif vs Rabi: statistically significant
- Kharif vs Zaid: statistically significant
- Rabi vs Zaid: statistically significant

Therefore, the yield distributions differ significantly between every pair of seasons.

However, the one-way ANOVA did not find a statistically significant difference in mean yield across seasons. This indicates that seasonal differences are more evident in the overall distribution and ranking of yields rather than in their mean values. The very small eta-squared value (η² = 0.0007) also indicates that season alone explains only a small proportion of the variation in yield.

## 17. Unusual Patterns and Key Findings

The analysis revealed several important patterns and variations in agricultural performance:

1. **Kharif showed the highest average profitability**, while Zaid recorded negative average profit.

2. **Seasonal mean yield differences were not statistically significant** according to the one-way ANOVA test (p = 0.232847).

3. **Yield distributions differed significantly across seasons**, as indicated by the Kruskal–Wallis test (p < 0.001). Pairwise Mann–Whitney U tests also showed significant differences between all three seasonal pairs after Bonferroni correction.

4. **Yield distributions were strongly right-skewed**, with the mean considerably higher than the median in all seasons and several high-yield outliers.

5. **Season alone explained very little of the variation in yield**, with an eta-squared value of 0.0007.

6. **Soil moisture varied noticeably across seasons**, with Kharif having the highest average soil moisture and Zaid the lowest, while soil pH and major nutrient levels remained relatively similar.

7. **Disease and pest risk was highest in Kharif**, although higher risk did not directly correspond to lower profitability.

8. **Yield and profit did not always follow the same pattern**, showing that market price, revenue, and production cost are also important when evaluating agricultural performance.

## 18. Key Insights

The major insights obtained from the analysis are:

- **Kharif was the best-performing season economically**, with the highest average profit and production among the three seasons.

- **Zaid showed the weakest economic performance**, with a negative average profit, indicating that cultivation during this season may involve higher economic challenges.

- **Seasonal average yield was not significantly different** according to ANOVA (p = 0.232847), although the yield distributions differed significantly according to the Kruskal–Wallis and pairwise Mann–Whitney tests.

- **Yield showed substantial variation within each season**, with several high-yield observations acting as outliers. This resulted in means being considerably higher than medians.

- **Season alone was not a strong explanation for yield variation**, as indicated by the very small eta-squared value of 0.0007.

- **Environmental conditions varied across seasons**, particularly rainfall and soil moisture, while soil pH and major nutrient levels remained relatively stable.

- **Kharif had the highest average disease and pest risk**, but this did not directly translate into lower profitability.

- **Crop selection strongly influenced economic performance.** The analysis showed that a crop with lower physical yield can still generate high profit when its market price and revenue are favorable.
